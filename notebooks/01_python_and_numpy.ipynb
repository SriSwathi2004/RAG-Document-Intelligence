## 3.1 Algebra: variables, functions, exponents, logarithms

```
import math
f = lambda x: 2 * x + 1          # a linear function (weight 2, bias 1)
print(f(3))
print(2 ** 3, math.log2(8), math.exp(0), round(math.exp(1), 4), round(math.log(math.e), 1))
# softmax turns scores into probabilities that sum to 1
scores = [2.0, 1.0, 0.1]
exps = [math.exp(s) for s in scores]
probs = [e / sum(exps) for e in exps]
print([round(p, 3) for p in probs], round(sum(probs), 3))
```

## 3.3 Dot product, magnitude, normalisation

```
import numpy as np
a = np.array([3.0, 4.0]); b = np.array([4.0, 3.0])
print(a @ b)                              # 3*4 + 4*3 = 24
print(np.linalg.norm(a))                  # sqrt(9+16) = 5
a_unit = a / np.linalg.norm(a)
print(a_unit, np.linalg.norm(a_unit))
```

## 3.4 Distances and cosine similarity (the heart of semantic search)

```
import numpy as np
q   = np.array([1.0, 1.0, 0.0])    # "question"
d1  = np.array([2.0, 2.0, 0.0])    # same direction, longer
d2  = np.array([0.0, 0.0, 3.0])    # unrelated
cos = lambda a, b: a @ b / (np.linalg.norm(a) * np.linalg.norm(b))
print(round(cos(q, d1), 3), round(cos(q, d2), 3))
print(round(np.linalg.norm(q - d1), 3), round(np.linalg.norm(q - d2), 3))
```

## 3.5 Probability and statistics

```
import numpy as np
scores = np.array([0.82, 0.74, 0.41, 0.12, 0.08])
print(scores.mean(), round(scores.var(), 4), round(scores.std(), 4))
# conditional probability from counts: P(relevant | score >= 0.5)
relevant = np.array([1, 1, 0, 0, 0]); high = scores >= 0.5
print(relevant[high].mean())                 # both high scores were relevant
print(relevant[~high].mean())
```

## 3.6 Calculus: derivatives, gradients, gradient descent

```
# minimise loss(w) = (w - 3)^2 ; derivative = 2 * (w - 3) ; best w is 3
w, lr = 0.0, 0.1
for step in range(1, 6):
    grad = 2 * (w - 3)
    w = w - lr * grad
    print(f"step {step}: w = {w:.4f}, loss = {(w - 3) ** 2:.4f}")
# too large a learning rate overshoots and diverges
w = 0.0
for _ in range(5):
    w = w - 1.1 * 2 * (w - 3)
print("lr=1.1 ->", round(w, 3))
```

## 4.2 Broadcasting, vectorisation, dot product, normalisation

```
import numpy as np
chunks = np.array([[3.0, 4.0], [1.0, 0.0], [0.0, 2.0]])   # 3 chunk embeddings
query = np.array([1.0, 1.0])
print(chunks - query)                              # broadcasting: (3,2) - (2,)
norms = np.linalg.norm(chunks, axis=1, keepdims=True)
chunks_n = chunks / norms                          # normalise each row
q_n = query / np.linalg.norm(query)
sims = chunks_n @ q_n                              # all cosine similarities at once
print(sims.round(3), "best chunk index:", int(sims.argmax()))
top2 = np.argsort(-sims)[:2]
print("top-2 indices:", top2)
```
