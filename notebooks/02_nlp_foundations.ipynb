## 5.1 Text preprocessing: normalisation, tokenisation, stop words

```
import re
text = "Machine learning is great. It learns from DATA! Is it magic? No."
sentences = re.split(r"(?<=[.!?])\s+", text)
print(sentences)
words = re.findall(r"[a-z]+", text.lower())
print(words)
stop = {"is", "it", "from", "no"}
print([w for w in words if w not in stop])
```

## 5.2 Stemming and lemmatisation

```
def toy_stem(w):
    for suf in ("ing", "ed", "es", "s"):
        if w.endswith(suf) and len(w) - len(suf) >= 3:
            return w[: -len(suf)]
    return w
print([toy_stem(w) for w in ["training", "trained", "trains", "tree", "studies"]])
```

## 5.3 Vocabulary, word frequency, n-grams

```
from collections import Counter
doc = "random forest uses many trees and a random forest reduces overfitting"
tokens = doc.split()
print(Counter(tokens).most_common(3))
bigrams = list(zip(tokens, tokens[1:]))
print(Counter(bigrams).most_common(2))
print(len(set(tokens)), "distinct tokens")
```

## 5.4 Bag of Words and TF-IDF

```
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
docs = ["random forest uses many decision trees",
        "gradient descent minimises the loss",
        "decision trees split the data"]
cv = CountVectorizer()
print(cv.fit_transform(docs).toarray())
print(cv.get_feature_names_out())
tv = TfidfVectorizer()
X = tv.fit_transform(docs)
row = dict(zip(tv.get_feature_names_out(), X[0].toarray()[0].round(2)))
print({k: v for k, v in row.items() if v > 0})
```

## 5.5 Text similarity and the limits of traditional NLP

```
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
corpus = ["The car is fast", "The automobile is quick", "The banana is yellow"]
X = TfidfVectorizer().fit_transform(corpus)
print(cosine_similarity(X).round(2))
```
